```python
import os
from glob import glob
import pandas as pd
```

```python
base_path = '../../../../../Downloads/HAM10000'
images_path = f"{base_path}/HAM10000_images"

metadata = pd.read_csv(f'{base_path}/HAM10000_metadata.csv')
metadata

images_path_crop = '../../../../../Downloads/HAM10000_BOUNDING_BOX_CROP'
```

```python
metadata[metadata['dx'] == 'nv']
```

```python
images = sorted(glob(os.path.join(images_path_crop,"*jpg")))
len(images)
```

```python
metadata = metadata[['lesion_id', 'image_id', 'dx']]
mask = (metadata['dx'] == 'mel') | (metadata['dx'] == 'nv')
metadata = metadata[mask]
metadata
```

```python
def get_classification(dx):
    if dx == 'mel':
        return 'Malignant'
    return 'Benignant'
```

```python
# Crear paths
for index, row in metadata.iterrows():
    full_path = f'{images_path_crop}/{row['image_id']}.jpg'
    metadata.at[index, 'path'] = full_path
    metadata.at[index, 'classification'] = get_classification(row['dx'])
```

```python
metadata.to_csv('../data/02_dataset_crop.csv', index=False)
```
